# 🧮 Matrix - Complete Theory

## 📘 Definition
A **Matrix** is a two-dimensional array of numbers arranged in rows and columns. It is widely used in mathematics, physics, computer science, and engineering for various operations and transformations.

---

## 🧱 Matrix Structure

A matrix with `m` rows and `n` columns is called an **m x n (m by n) matrix**.

```text
A = | a11  a12  a13 |
    | a21  a22  a23 |
    | a31  a32  a33 |
```

Here, A is a 3×3 matrix.

---

## 🧩 Types of Matrices

| Type                 | Description                                                             |
|----------------------|-------------------------------------------------------------------------|
| **Row Matrix**        | A matrix with only one row.                                             |
| **Column Matrix**     | A matrix with only one column.                                          |
| **Square Matrix**     | Number of rows = number of columns (n x n).                             |
| **Diagonal Matrix**   | Square matrix where all off-diagonal elements are zero.                 |
| **Scalar Matrix**     | Diagonal matrix with all diagonal elements equal.                       |
| **Identity Matrix**   | Square matrix with 1s on the diagonal and 0s elsewhere.                 |
| **Zero/Null Matrix**  | All elements are 0.                                                     |
| **Symmetric Matrix**  | Aᵀ = A; matrix equal to its transpose.                                  |
| **Skew-Symmetric**    | Aᵀ = -A; transpose equals negative of matrix.                           |
| **Upper/Lower Triangular** | All elements below (or above) the diagonal are 0.              |

---

## 🔁 Matrix Operations

| Operation              | Description                                                                 | Time Complexity |
|------------------------|-----------------------------------------------------------------------------|-----------------|
| **Addition**           | Element-wise addition of two matrices of same dimension.                   | O(m×n)          |
| **Subtraction**        | Element-wise subtraction.                                                   | O(m×n)          |
| **Scalar Multiplication** | Multiply each element by a scalar.                                     | O(m×n)          |
| **Matrix Multiplication** | Dot product between rows and columns.                                   | O(n³) (naive)   |
| **Transpose**          | Flip matrix over its diagonal.                                              | O(m×n)          |
| **Determinant**        | Only for square matrices.                                                   | O(n!) (general) |
| **Inverse**            | Exists only if matrix is non-singular.                                      | O(n³)           |

---

## ✅ Properties of Matrix Multiplication

- Not commutative: **A × B ≠ B × A**
- Associative: **(A × B) × C = A × (B × C)**
- Distributive: **A × (B + C) = A × B + A × C**

---

## ⚠️ Conditions for Operations

- **Addition/Subtraction**: Both matrices must be of the **same dimension**.
- **Multiplication**: Columns of A must match rows of B.
- **Inverse**: Only for **square matrices** with **non-zero determinant**.

---

## 🧠 Applications of Matrices

- 🔢 **Solving linear equations**
- 🎮 **Computer graphics and 3D transformations**
- 📊 **Data representation and manipulation**
- 🧠 **Neural networks and machine learning**
- 🔐 **Cryptography and encoding**
- 📷 **Image processing**

---

## 💡 Real-World Examples

- 💻 Pixel data in images
- 🧠 Weight matrices in neural networks
- 🌐 Graph adjacency matrices
- 🛰️ Transformations in robotics and aerospace
- 🔒 Encoding/decoding messages securely

---

## 🔚 Summary

Matrices are a foundational structure in linear algebra and computational systems. Their powerful properties and diverse operations enable a wide range of applications from simple arithmetic to advanced machine learning.

````python
class Node(object):
    """Node class for a singly linked list."""
    def __init__(self, data=None,next=None):
        """Initialize a node with data and next pointer."""
        self.data = data
        self.next = None

class LinkedList():
    """Singly linked list class."""
    def __init__(self):
        """Initialize an empty linked list."""
        self.head = None

    def __len__(self):
        """Return the length of the linked list."""
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count


    def inser_at_beginning(self, data):
        """Insert a new node at the beginning of the list."""
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def insert_at_end(self, data):
        """Insert a new node at the end of the list."""
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        last_node = self.head
        while last_node.next:
            last_node = last_node.next
        last_node.next = new_node

    def insert_at_position(self, data, position):
        """Insert a new node at a specified position in the list.""" 
        if position < 0 or position > len(self):
            raise IndexError("Index out of range")
        if position == 0:
            self.inser_at_beginning(data)
            return
        new_node = Node(data)
        current = self.head
        for _ in range(position - 1):
            current = current.next
        new_node.next = current.next
        current.next = new_node

    def delete_node(self, key): 
        """Delete the first occurrence of a node with the given key."""
        if self.head is None:   
            return
        if self.head.data == key:
            self.head = self.head.next
            return
        current = self.head 
        while current.next:
            if current.next.data == key:
                current.next = current.next.next
                return
            current = current.next
        raise ValueError("Node with data {} not found".format(key)) 
    
    def search(self, key):
        """Search for a node with the given key."""
        current = self.head
        while current:
            if current.data == key:
                return True
            current = current.next
        return False
    
    def clear(self):
        """Clear the linked list."""
        self.head = None
        self.tail = None
        self.size = 0
        self.current = None
        self.index = 0

    def is_empty(self):
        """Check if the linked list is empty."""
        return self.head is None
        
    def get_size(self):
        """Return the size of the linked list."""
        return len(self)
    
    def get_head(self):
        """Return the head node of the linked list."""
        return self.head.data if self.head else None    
    
    def get_tail(self):
        """Return the tail node of the linked list."""
        current = self.head
        while current and current.next:
            current = current.next
        return current.data if current else None    
    

    def traverse(self):
        """Traverse the list and print the data in each node."""
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")

if __name__=="__main__":
    # Create a new linked list
    ll = LinkedList()

    # Insert nodes at the beginning
    ll.inser_at_beginning(3)
    ll.inser_at_beginning(2)
    ll.inser_at_beginning(1)
    # Traverse the list and print the nodes
    ll.traverse()

    # Insert nodes at the end
    ll.insert_at_end(4)
    # Traverse the list and print the nodes
    ll.traverse()

    # Insert nodes at a specific position
    ll.insert_at_position(5, 2)
    ll.traverse()

    # Delete a node
    ll.delete_node(3)
    ll.traverse()

    # Search for a node
    print("Search for 4:", ll.search(4))

    print("Search for 3:", ll.search(3))

    # Clear the list
    # ll.clear()
    ll.traverse()
    print("List cleared.")
    print("Is list empty?", ll.is_empty())
    print("List size:", ll.get_size())
    print("List head:", ll.get_head())
    print("List tail:", ll.get_tail())

    
````

# 🔗 Doubly Linked List (DLL)

## 📘 Definition
A **Doubly Linked List (DLL)** is a type of linked list where each node contains **three parts**:
- `data`: The actual value.
- `prev`: Pointer to the **previous** node.
- `next`: Pointer to the **next** node.

This bidirectional structure allows traversal in both forward and backward directions.

---

## 🧩 Structure of a Node

+-------+--------+-------+ | prev | data | next | +-------+--------+-------+


### Example in Python-style pseudocode:
```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


## ⚙️ Operations and Their Complexity

| Operation                  | Description                                                                 | Time Complexity |
|---------------------------|-----------------------------------------------------------------------------|-----------------|
| **Traversal (Forward)**   | Start from `head` and follow `next` links until `None`.                     | O(n)            |
| **Traversal (Backward)**  | Start from `tail` and follow `prev` links until `None`.                     | O(n)            |
| **Insertion at Head**     | Update `next` of new node and `prev` of current head, then update head.     | O(1)            |
| **Insertion at Tail**     | If `tail` pointer exists, insert directly. Else, traverse to the end.       | O(1) / O(n)     |
| **Insertion After Node**  | Update four pointers: `node.next`, `new.prev`, `new.next`, `next.prev`.     | O(1)            |
| **Deletion (Given Node)** | Update `prev.next` and `next.prev` to skip the node to be deleted.          | O(1)            |
| **Search**                | Traverse through the list and compare each `data`.                          | O(n)            |


````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0
        self.current = None
        self.index = 0

    def insert_at_end(self, data):
        node = Node(data)
        if self.head is None:
            self.head = node    
            self.tail = node
        else:
            self.tail.next = node
            node.prev = self.tail
            self.tail = node
        self.size += 1
        self.current = self.tail
        self.index = self.size - 1

    def insert_at_beginning(self, data):
        node = Node(data)
        if self.head is None:
            self.head = node
            self.tail = node
        else:
            node.next = self.head
            self.head.prev = node
            self.head = node
        self.size += 1
        self.current = self.head
        self.index = 0

    def insert_at_position(self, data, position):
        if position < 0 or position > self.size:
            raise IndexError("Index out of range")
        if position == 0:
            self.insert_at_beginning(data)
            return
        if position == self.size:
            self.insert_at_end(data)
            return
        node = Node(data)
        current = self.head
        for _ in range(position - 1):
            current = current.next
        node.next = current.next
        node.prev = current
        current.next.prev = node
        current.next = node
        self.size += 1
        self.current = node
        self.index = position

    def delete_node(self, data):
        current = self.head
        while current:
            if current.data == data:
                if current.prev:
                    current.prev.next = current.next
                else:
                    self.head = current.next
                if current.next:
                    current.next.prev = current.prev
                else:
                    self.tail = current.prev
                self.size -= 1
                return
            current = current.next
        raise ValueError("Node with data {} not found".format(data))
    
    def traverse(self):
        current = self.head
        while current:
            print(current.data, end=" <-> ")
            current = current.next
        print("None")

    def traverse_reverse(self):
        current = self.tail
        while current:
            print(current.data, end=" <-> ")
            current = current.prev
        print("None")

    def search(self, data):
        current = self.head
        while current:
            if current.data == data:
                return True
            current = current.next
        return False
    
    def clear(self):
        self.head = None
        self.tail = None
        self.size = 0
        self.current = None
        self.index = 0

    def is_empty(self):
        return self.size == 0
    
    def get_size(self):
        return self.size
    
    def get_head(self):
        return self.head.data if self.head else None
    
    def get_tail(self):
        return self.tail.data if self.tail else None
    
    def get_index(self):
        return self.index if self.current else -1
    
if __name__=="__main__":
    # Create a new doubly linked list
    dll = DoublyLinkedList()

    # Insert nodes at the beginning
    dll.insert_at_beginning(3)
    dll.insert_at_beginning(2)
    dll.insert_at_beginning(1)
    # Traverse the list and print the nodes
    dll.traverse()

    # Insert nodes at the end
    dll.insert_at_end(4)
    # Traverse the list and print the nodes
    dll.traverse()

    # Insert nodes at a specific position
    dll.insert_at_position(5, 2)
    dll.traverse()

    # Delete a node
    dll.delete_node(3)
    dll.traverse()

    # Search for a node
    print("Search for 4:", dll.search(4))

    print("Search for 3:", dll.search(3))

    # Clear the list
    # dll.clear()
    dll.traverse()
````

# Circular Linked List 

# 🔁 Circular Linked List (CLL)

## 📘 Definition
A **Circular Linked List** is a variation of the linked list in which the last node **points back to the first node**, forming a circular loop. This makes the list **cyclical** instead of terminating with `null` or `None`.

---

## 🧩 Types of Circular Linked Lists

### 1. 🔘 **Singly Circular Linked List**
- Each node has a `data` field and a `next` pointer.
- The `next` of the last node points to the **head**.

```text
head → Node1 → Node2 → Node3
          ↑               ↓
          ← ← ← ← ← ← ← ←
```

### 2. 🌀 **Doubly Circular Linked List**
- Each node contains `data`, `prev`, and `next`.
- The `next` of the last node points to the **head**.
- The `prev` of the head points to the **last** node.

```text
head ↔ Node1 ↔ Node2 ↔ Node3 ↔ head
```

---

## 🧠 Characteristics

- No node has a `null` next (or `prev`), making the list **circular**.
- Can be **traversed infinitely** unless explicitly stopped.
- Useful for applications requiring **looped structures** or **continuous cycling**.
- Traversal must use a **stopping condition** (e.g., `while current != head:`).

---

## ⚙️ Operations and Their Complexity

| Operation                  | Description                                                                                   | Time Complexity |
|---------------------------|-----------------------------------------------------------------------------------------------|-----------------|
| **Traversal (Forward)**   | Start from head, stop when back at head.                                                     | O(n)            |
| **Insertion at Head**     | Insert before head and update last node’s next.                                               | O(n) / O(1)*    |
| **Insertion at Tail**     | Insert after last node and point to head.                                                     | O(n) / O(1)*    |
| **Insertion After Node**  | Add node after a specific node.                                                               | O(1)            |
| **Deletion (Given Node)** | Update previous node’s next to skip deleted node. Handle special cases if deleting head.     | O(n)            |
| **Search**                | Traverse until back at head or data found.                                                    | O(n)            |

\* If tail pointer is maintained.

---

## ✅ Advantages

- 🔄 **Efficient cycling** through elements, ideal for repetitive looping.
- 🧠 **No null checks** at end, easier to wrap around from end to beginning.
- 🔂 Useful in applications requiring round-robin behavior.

---

## ⚠️ Disadvantages

- 🚧 **Harder to implement** than linear lists (requires loop-breaking conditions).
- 🧮 **More difficult to debug** due to infinite loops if not carefully handled.
- 🧠 **Memory overhead** in doubly circular lists (due to extra `prev` pointer).

---

## 💡 Use Cases / Real-World Examples

- 🎵 **Media players** – Playing a playlist in a loop.
- 🎯 **Round-robin scheduling** – CPU scheduling or fair task execution.
- 🔄 **Buffer management** – Circular buffers or queues.
- 🧩 **Games** – Turn-based games where turns rotate among players.
- ⌚ **Clock-based data structures** – Algorithms where access wraps around (e.g., page replacement algorithms in OS).

---

## 🔚 Summary

A **Circular Linked List** is a powerful variation of a linked list suited for problems involving **cyclic traversal**. It enables smooth looping over data without needing to restart manually and is well-suited for scheduling, buffer management, and repetitive control flows.


````python
# Circular Linked List
# 1. Create a node class
# 2. Create a circular linked list class
# 3. Create a method to insert a node at the end of the list
# 4. Create a method to insert a node at the beginning of the list
# 5. Create a method to insert a node at a specific position in the list
# 6. Create a method to delete a node from the list
# 7. Create a method to search for a node in the list
# 8. Create a method to traverse the list and print the nodes
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class CircularLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0
        self.current = None
        self.index = 0

    def insert_at_end(self, data):
        node = Node(data)
        if self.head is None:
            self.head = node
            self.tail = node
            node.next = node
            node.prev = node
        else:
            node.prev = self.tail
            node.next = self.head
            self.tail.next = node
            self.head.prev = node
            self.tail = node
        self.size += 1
        self.current = self.tail
        self.index = self.size - 1
    
    def insert_at_beginning(self, data):    
        node = Node(data)
        if self.head is None:
            self.head = node    
            self.tail = node
            node.next = node
            node.prev = node    
        else:
            node.next = self.head
            node.prev = self.tail
            self.tail.next = node
            self.head.prev = node
            self.head = node
        self.size += 1
        self.current = self.head    
        self.index = 0

    def insert_at_position(self, data, position):
        if position < 0 or position > self.size:
            raise IndexError("Index out of range")
        if position == 0:
            self.insert_at_beginning(data)
            return
        if position == self.size:
            self.insert_at_end(data)
            return
        node = Node(data)
        current = self.head
        for _ in range(position - 1):
            current = current.next
        node.next = current.next
        node.prev = current
        current.next.prev = node
        current.next = node
        self.size += 1
        self.current = node
        self.index = position

    def delete_node(self, data):    
        current = self.head
        while current:
            if current.data == data:
                if current.prev:
                    current.prev.next = current.next
                else:
                    self.head = current.next
                if current.next:
                    current.next.prev = current.prev
                else:
                    self.tail = current.prev
                self.size -= 1
                return
            current = current.next
        raise ValueError("Node with data {} not found".format(data))
    
    def search(self, data):
        current = self.head
        if current is None:
            return False
        while True:
            if current.data == data:
                return True
            current = current.next
            if current == self.head:
                break
        return False
    
    def traverse(self):
        current = self.head
        if current is None:
            return
        while True:
            print(current.data, end=" -> ")
            current = current.next
            if current == self.head:
                break
        print("None")

    def traverse_reverse(self):
        current = self.tail
        if current is None:
            return
        while True:
            print(current.data, end=" -> ")
            current = current.prev
            if current == self.tail:
                break
        print("None")

    def clear(self):
        self.head = None
        self.tail = None
        self.size = 0
        self.current = None
        self.index = 0

    def is_empty(self):
        return self.size == 0
    
    def get_size(self):
        return self.size
    

if __name__=="__main__":
    # Create a new circular linked list
    cll = CircularLinkedList()

    # Insert nodes at the beginning
    cll.insert_at_beginning(3)
    cll.insert_at_beginning(2)
    cll.insert_at_beginning(1)
    # Traverse the list and print the nodes
    cll.traverse()

    # Insert nodes at the end
    cll.insert_at_end(4)
    # Traverse the list and print the nodes
    cll.traverse()

    # Insert nodes at a specific position
    cll.insert_at_position(5, 2)
    cll.traverse()

    # Delete a node
    cll.delete_node(3)
    cll.traverse()

    # Search for a node
    print("Search for 4:", cll.search(4))

    print("Search for 3:", cll.search(3))

    # Clear the list
    # cll.clear()
````

# 📚 Stack Data Structure – Complete Theory

| `isEmpty()` | Check if stack is empty          | O(1)            |

````python
# Queue using array
class Queue:

    def __init__(self, max_size):
        self.max_size = max_size
        self.queue = [None] * max_size
        self.front = self.rear = -1

    def is_empty(self): 
        return self.front == -1
    
    def is_full(self):
        return self.rear == self.max_size - 1
        
    def is_empty(self):
        return self.front == -1
    
    def enqueue(self, data):
        if self.is_full():
            print("Queue is full")
            return
        if self.is_empty():
            self.front = 0
        self.rear += 1
        self.queue[self.rear] = data
        print(f"Enqueued {data} to queue")

    def dequeue(self):  
        if self.is_empty():
            print("Queue is empty")
            return None
        data = self.queue[self.front]
        if self.front == self.rear:
            self.front = self.rear = -1
        else:
            self.front += 1
        return data
    
    def peek(self):
        if self.is_empty():
            print("Queue is empty")
            return None
        return self.queue[self.front]
    
    def size(self):
        if self.is_empty():
            return 0
        return self.rear - self.front + 1

    def display(self):
        if self.is_empty():
            print("Queue is empty")
            return
        print("Queue (front -> rear):", end=" ")
        for i in range(self.front, self.rear + 1):
            print(self.queue[i], end=" -> ")
        print("None")   

        

if __name__=="__main__":
    queue = Queue(5)
    queue.enqueue(1)
    queue.enqueue(2)
    queue.enqueue(3)
    queue.display()
    print("Dequeued element:", queue.dequeue())
    queue.display()
    print("Size of queue:", queue.size())
    print(queue.is_empty())
````

````python
# Queue using linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class QueueLL:
    def __init__(self):
        self.front = None
        self.rear = None

        
    def enqueue(self, data):
        new_node = Node(data)
        if self.rear is None:
            self.front = self.rear = new_node
            return
        self.rear.next = new_node
        self.rear = new_node
    def dequeue(self):
        if self.front is None:
            return None
        temp = self.front
        self.front = self.front.next
        if self.front is None:
            self.rear = None
        return temp.data
    def peek(self):
        if self.front is None:
            return None
        return self.front.data
    def is_empty(self):
        return self.front is None
    def size(self):
        count = 0
        current = self.front
        while current:
            count += 1
            current = current.next
        return count
    def display(self):
        if self.is_empty():
            print("Queue is empty")
            return
        current = self.front
        print("Queue (front -> rear):", end=" ")
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
    def clear(self):
        self.front = None
        self.rear = None
        print("Queue cleared")  

    def search(self, data):
        current = self.front 
        while current:
            if current.data == data:
                return True
            current = current.next
        return False
    def reverse(self):
        prev = None
        current = self.front
        self.rear = self.front
        while current:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node
        self.front = prev
    def to_list(self):
        result = []
        current = self.front
        while current:
            result.append(current.data)
            current = current.next
        return result
    def from_list(self, lst):
        self.clear()
        for item in lst:
            self.enqueue(item)
    def get_front(self):
        return self.front.data if self.front else None
    def get_rear(self):
        return self.rear.data if self.rear else None
    def get_size(self):
        return self.size
    
if __name__=="__main__":
    queue = QueueLL()
    queue.enqueue(1)
    queue.enqueue(2)
    queue.enqueue(3)
    queue.display()
    print("Dequeued element:", queue.dequeue())
    queue.display()
    print("Size of queue:", queue.size())    
````

````python
class Node:

    def __init__(self, data, parent):
        self.data = data
        self.parent = parent
        self.right_node = None
        self.left_node = None


class BinarySearchTree:

    def __init__(self):
        self.root = None

    def remove(self, data):
        if self.root:
            self.remove_node(data, self.root)

    def insert(self, data):
        if self.root is None:
            self.root = Node(data, None)
        else:
            self.insert_node(data, self.root)

    def insert_node(self, data, node):
        # we have to go to the left subtree
        if data < node.data:
            if node.left_node:
                self.insert_node(data, node.left_node)
            else:
                node.left_node = Node(data, node)
        # we have to visit the right subtree
        else:
            if node.right_node:
                self.insert_node(data, node.right_node)
            else:
                node.right_node = Node(data, node)

    def remove_node(self, data, node):

        if node is None:
            return

        if data < node.data:
            self.remove_node(data, node.left_node)
        elif data > node.data:
            self.remove_node(data, node.right_node)
        else:

            if node.left_node is None and node.right_node is None:
                print("Removing a leaf node...%d" % node.data)

                parent = node.parent

                if parent is not None and parent.left_node == node:
                    parent.left_node = None
                if parent is not None and parent.right_node == node:
                    parent.right_node = None

                if parent is None:
                    self.root = None

                del node

            elif node.left_node is None and node.right_node is not None:  # node !!!
                print("Removing a node with single right child...")

                parent = node.parent

                if parent is not None:
                    if parent.left_node == node:
                        parent.left_node = node.left_node
                    if parent.rightChild == node:
                        parent.right_node = node.right_node
                else:
                    self.root = node.right_node

                node.right_node.parent = parent
                del node

            elif node.right_node is None and node.left_node is not None:
                print("Removing a node with single left child...")

                parent = node.parent

                if parent is not None:
                    if parent.left_node == node:
                        parent.left_node = node.left_node
                    if parent.right_node == node:
                        parent.right_node = node.left_node
                else:
                    self.root = node.left_node

                node.left_node.parent = parent
                del node

            else:
                print('Removing node with two children....')

                predecessor = self.get_predecessor(node.left_node)

                temp = predecessor.data
                predecessor.data = node.data
                node.data = temp

                self.remove_node(data, predecessor)

    def get_predecessor(self, node):
        if node.right_node:
            return self.get_predecessor(node.right_node)

        return node

    def traverse(self):
        if self.root is not None:
            self.traverse_in_order(self.root)

    def traverse_in_order(self, node):

        if node.left_node:
            self.traverse_in_order(node.left_node)

        print(node.data)

        if node.right_node:
            self.traverse_in_order(node.right_node)
````

def naive_search(text, pattern):
    n, m = len(text), len(pattern)
    positions = []
    for i in range(n - m + 1):
        if text[i:i+m] == pattern:
            positions.append(i)
    return positions

def rabin_karp(text, pattern, d=256, q=101):
    n, m = len(text), len(pattern)
    h = pow(d, m-1, q)
    p = t = 0
    positions = []
    
    # Hash calculation
    for i in range(m):
        p = (d*p + ord(pattern[i])) % q
        t = (d*t + ord(text[i])) % q
    
    # Pattern matching
    for i in range(n-m+1):
        if p == t and text[i:i+m] == pattern:
            positions.append(i)
        if i < n-m:
            t = (d*(t-ord(text[i])*h) + ord(text[i+m])) % q
    return positions
